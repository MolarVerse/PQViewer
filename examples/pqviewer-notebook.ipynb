# PQViewer in Jupyter

Start the real local PQViewer application in a notebook cell. The iframe keeps the interactive 3D renderer, structure and cell editing, command search, selections, representations, and figure export.

## Setup

Install the notebook helper and optional ASE readers with:

```bash
python -m pip install 'MolarVerse-PQViewer[jupyter,ase]'
```

## Open a structure

The returned object owns a local server and displays itself as an iframe. This example uses the centered SrTiO3 perovskite fixture.

````python
from pathlib import Path
from pqviewer import view

source = next(
    path for path in (
        Path("examples/strontium-titanate.extxyz"),
        Path("strontium-titanate.extxyz"),
    )
    if path.exists()
)
viewer = view(source, height=620, title="SrTiO3 perovskite")
viewer
````

Press `/` or use the visible **Search** control, then choose **Representation · Polyhedra** to inspect the TiO6 octahedra.

````python
import json
from urllib.request import urlopen

with urlopen(f"{viewer.url}api/health", timeout=5) as response:
    status = json.load(response)

assert status == {"status": "ok"}
print(f"PQViewer is ready at {viewer.url}")
````

Call `viewer.close()` when the embedded viewer is no longer needed. Local notebook kernels work directly; remote kernels require the notebook environment to forward the viewer port.
